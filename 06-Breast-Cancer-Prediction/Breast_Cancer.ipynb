```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
```

```python
brcn=pd.read_csv('https://raw.githubusercontent.com/SinhaPushkar047/ML_Learning_Projects/refs/heads/main/06-Breast-Cancer-Prediction/data.csv').drop(columns=['id','Unnamed: 32'])
```

```python
pd.set_option("display.max_columns",None)
```

# **Data Encoding**

```python
lc=LabelEncoder()
brcn['diagnosis']=lc.fit_transform(brcn['diagnosis'])
```

```python
brcn
```

# **Train Test Split**

```python
x_train,x_test,y_train,y_test=train_test_split(brcn.drop(columns='diagnosis'),brcn['diagnosis'],test_size=0.2,random_state=42)
```

# **PCA**

```python
scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(x_train)

pca_full = PCA()
pca_full.fit(X_train_scaled)

cumulative_variance = np.cumsum(
    pca_full.explained_variance_ratio_
)

# Plot

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")
plt.grid()

plt.show()
```

```python
threshold = 0.99

n_components_99 = np.argmax(
    cumulative_variance >= threshold
) + 1

print(
    "\nComponents required for 99% variance:",
    n_components_99
)
```

# **KNN**

```python
pipe=Pipeline([
    ('scaling',MinMaxScaler()),
    ('knn',KNeighborsClassifier())
])
```

```python
pipe2=Pipeline([
    ('scaling',MinMaxScaler()),
    ('pca',PCA(n_components=16)),
    ('knn',KNeighborsClassifier())
])
```

```python
n_neighbors=[5,10,15,20,25]
param_grid={'knn__n_neighbors':n_neighbors}
```

```python
grid=GridSearchCV(
    estimator=pipe2,
    param_grid=param_grid,
    cv=5,verbose=2,
    scoring='f1',
    n_jobs=-1
)
```

```python
grid.fit(x_train,y_train)
```

```python
y_pred=grid.predict(x_test)
```

```python
print(classification_report(y_test,y_pred))
```
